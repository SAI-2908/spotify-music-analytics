```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import os
os.listdir()
```

```python
df_clean = pd.read_csv("spotify_data clean.csv")
df_final = pd.read_csv("track_data_final.csv")

print("Clean dataset:", df_clean.shape)
print("Final dataset:", df_final.shape)

display(df_clean.head())
display(df_final.head())
```

```python
print("Clean dataset:", df_clean.shape)
print("Final dataset:", df_final.shape)

print("\nCLEAN DATASET COLUMNS:")
print(df_clean.columns.tolist())

print("\nFINAL DATASET COLUMNS:")
print(df_final.columns.tolist())
```

```python
print(df_clean.isnull().sum())
print("\nDuplicate rows:", df_clean.duplicated().sum())
print("\nData types:")
print(df_clean.dtypes)
```

```python
print("Release date examples:")
print(df_clean["album_release_date"].head(10))

print("\nPopularity ranges:")
print(df_clean[["track_popularity", "artist_popularity"]].describe())

print("\nAlbum types:")
print(df_clean["album_type"].value_counts())

print("\nMissing artist names:")
display(df_clean[df_clean["artist_name"].isna()])
```

```python
df_clean = df_clean.dropna(subset=["artist_name"]).copy()

df_clean["artist_genres"] = df_clean["artist_genres"].fillna("Unknown")

df_clean["album_release_date"] = pd.to_datetime(
    df_clean["album_release_date"], errors="coerce"
)

df_clean["release_year"] = df_clean["album_release_date"].dt.year

print("Dataset shape:", df_clean.shape)
print("Missing values:\n", df_clean.isnull().sum())
```

```python
print("Release year range:")
print(df_clean["release_year"].min(), "-", df_clean["release_year"].max())

print("\nTracks per year:")
print(df_clean["release_year"].value_counts().sort_index().tail(10))

print("\nTop 10 most popular tracks:")
display(
    df_clean.nlargest(10, "track_popularity")[
        ["track_name", "artist_name", "track_popularity", "album_release_date"]
    ]
)
```

```python
#Number of tracks released each year
year_counts = df_clean["release_year"].value_counts().sort_index()

plt.figure(figsize=(12, 5))
year_counts.plot(kind="line", marker="o")
plt.title("Number of Tracks by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Tracks")
plt.grid(alpha=0.3)
plt.show()
```

```python
#Distribution of track popularity
plt.figure(figsize=(10, 5))
sns.histplot(df_clean["track_popularity"], bins=20, kde=True)
plt.title("Distribution of Spotify Track Popularity")
plt.xlabel("Track Popularity")
plt.ylabel("Number of Tracks")
plt.show()
```

```python
top_artists = (
    df_clean.groupby("artist_name")
    .agg(
        avg_popularity=("track_popularity", "mean"),
        track_count=("track_name", "count")
    )
    .query("track_count >= 5")
    .sort_values("avg_popularity", ascending=False)
    .head(10)
    .sort_values("avg_popularity")
)

print(top_artists)

top_artists["avg_popularity"].plot(
    kind="barh", figsize=(10, 6)
)
plt.title("Top 10 Artists by Average Track Popularity")
plt.xlabel("Average Track Popularity")
plt.ylabel("Artist")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_clean,
    x="track_duration_min",
    y="track_popularity",
    alpha=0.4
)

plt.title("Track Duration vs Spotify Popularity")
plt.xlabel("Track Duration (minutes)")
plt.ylabel("Track Popularity")
plt.tight_layout()
plt.show()

print("Correlation:", round(
    df_clean["track_duration_min"].corr(
        df_clean["track_popularity"]
    ), 3
))
```
